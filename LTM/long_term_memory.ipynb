### Creating Memories: A Guide to Preserving Your Precious Moments

```python
from langgraph.store.memory import InMemoryStore
```

```python
store = InMemoryStore()
```

```python
namespace = ('user','u1')
```

```python
store.put(namespace,key='1',value={'data':'user likes pizza'})
store.put(namespace,key='2',value={'data':'user prefers dark mode'})
```

```python
namespace2 = ('user','u2')
```

```python
store.put(namespace2,key='1',value={'data':'user likes paasta'})
store.put(namespace2,key='2',value={'data':'user prefers grid style navigation'})
```

#### Retrieving Memeories

```python
print(store.get(namespace,'1').value)   # type:ignore
print(store.get(namespace,'2').value)   # type:ignore
print(store.get(namespace2,'1').value)  # type:ignore
print(store.get(namespace2,'2').value)  # type:ignore
```

#### Retreiving All the Memories

```python
items = store.search(namespace2)

for it in items:
    print(it.value)
```

#### Semantic Search

```python
from langchain_openai import OpenAIEmbeddings
from dotenv import load_dotenv
```

```python
load_dotenv()
model = OpenAIEmbeddings(model="text-embedding-3-small")
```

```python
store = InMemoryStore(index = {'embed':model, 'dims':1536})
```

```python
namespace3 = ('user','u1')
```

```python
store.put(namespace3,'1',{'data':"user prefer consise answers over long explanations"})
store.put(namespace3,'2',{'data':"user user likes examples in python"})
store.put(namespace3,'3',{'data':"user usually works late at night"})
store.put(namespace3,'4',{'data':"user usually prefers dark mode in applications"})
store.put(namespace3,'5',{'data':"user is learning about langchain and langgraph"})
store.put(namespace3,'6',{'data':"user dislikes overly thechnical explanations and prefers simple language"})
store.put(namespace3,'7',{'data':"user prefers step by step reasoning and examples"})
store.put(namespace3,'8',{'data':"user is based in Uk"})
store.put(namespace3,'9',{'data':"user like real world analogies and examples to understand concepts"})
store.put(namespace3,'10',{'data':"user prefers bullet points and concise summaries over long paragraphs"})
```

```python
items = store.search(namespace3, query='what is the user currently learning', limit=3)
for item in items:
    print(item.value)
```

```python
items = store.search(namespace3, query="what are user's prefrences", limit=3)
for item in items:
    print(item.value)
```
